```python
import pytesseract
import cv2
from google.colab.patches import cv2_imshow

def tesseract_predict(image):
  bboxes = pytesseract.image_to_boxes(image)
  content = pytesseract.image_to_string(image)
  h, w, _ = image.shape
  for b in bboxes.splitlines():
    b = b.split(" ")
    text, xmin, ymin, xmax, ymax, name = b
    ymin, ymax = h - int(ymin), h - int(ymax)
    cv2.rectangle(image, (int(xmin), int(ymin)), (int(xmax), int(ymax)), (0, 0, 255),1, cv2.LINE_AA)
  return image, content


```

```python
image = "/content/Can-cuoc-cong-dan-bao-lau-het-han-theo-quy-dinh-nam-2022-1.jpg"
image = cv2.imread(image)
image, content = tesseract_predict(image)
```
