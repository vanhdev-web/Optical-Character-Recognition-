```python
import easyocr
import cv2
from google.colab.patches import cv2_imshow

def easyocr_predict(image):
  reader = easyocr.Reader(["vi","en"])
  output = reader.readtext(image)
  contents = []
  for item in output:
    bbox, content, score = item
    contents.append(content)
    min_point, _,max_point,_ = bbox
    cv2.rectangle(image, min_point, max_point, (0, 255, 0), 2)
  return image, contents


```

```python
image_path = "/content/Can-cuoc-cong-dan-bao-lau-het-han-theo-quy-dinh-nam-2022-1.jpg"
image = cv2.imread(image_path)

image, contents = easyocr_predict(image)
cv2_imshow(image)
for content in contents:
  print(content)
```
